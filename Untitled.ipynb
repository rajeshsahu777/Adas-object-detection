```python
import os, shutil, random, cv2
from PIL import Image
import torch
import torchvision.transforms as transforms
import torchvision.models as models
from ultralytics import YOLO

# -----------------------------
# Paths
# -----------------------------
src_images = r"D:\adas\data_object_image_2\training\image_2"
dst_images = r"D:\adas\images"
dst_labels = r"D:\adas\labels"

# -----------------------------
# Step 1: Create YOLO folders
# -----------------------------
for split in ["train", "val", "test"]:
    os.makedirs(os.path.join(dst_images, split), exist_ok=True)
    os.makedirs(os.path.join(dst_labels, split), exist_ok=True)

# -----------------------------
# Step 2: Split images
# -----------------------------
images = [f for f in os.listdir(src_images) if f.endswith(".png")]
random.shuffle(images)

train_split = int(0.7 * len(images))
val_split   = int(0.15 * len(images))

train_imgs = images[:train_split]
val_imgs   = images[train_split:train_split+val_split]
test_imgs  = images[train_split+val_split:]

def copy_files(file_list, split):
    for f in file_list:
        shutil.copy(os.path.join(src_images, f), os.path.join(dst_images, split, f))

copy_files(train_imgs, "train")
copy_files(val_imgs, "val")
copy_files(test_imgs, "test")

print("✅ Split complete:", len(train_imgs), "train,", len(val_imgs), "val,", len(test_imgs), "test")

# -----------------------------
# Step 3: Auto-label using YOLO proposals + MobileNetV3 classifier
# -----------------------------
# Load YOLO for bounding box proposals
yolo = YOLO("yolov8n.pt")

# Load MobileNetV3 classifier
classifier = models.mobilenet_v3_small(pretrained=True)
classifier.classifier[3] = torch.nn.Linear(classifier.classifier[3].in_features, 12)  # 12 classes
classifier.eval()

transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor()
])

# Define Indian road classes
classes = [
    "Car","Bus","Truck","Van","Auto_rickshaw","Motorbike",
    "Bicycle","Pedestrian","Animal","Traffic_light","Traffic_sign","Misc"
]

def auto_label_images(img_dir, label_dir):
    for img_file in os.listdir(img_dir):
        if not img_file.endswith(".png"): continue
        img_path = os.path.join(img_dir, img_file)
        results = yolo(img_path)   # get bounding boxes
        yolo_lines = []
        for box in results[0].boxes.xyxy:
            xmin, ymin, xmax, ymax = map(int, box.tolist())
            roi = cv2.imread(img_path)[ymin:ymax, xmin:xmax]
            if roi.size == 0: continue
            roi_tensor = transform(Image.fromarray(roi)).unsqueeze(0)
            pred = classifier(roi_tensor).argmax().item()
            x_center = (xmin + xmax) / 2 / results[0].orig_shape[1]
            y_center = (ymin + ymax) / 2 / results[0].orig_shape[0]
            width    = (xmax - xmin) / results[0].orig_shape[1]
            height   = (ymax - ymin) / results[0].orig_shape[0]
            yolo_lines.append(f"{pred} {x_center} {y_center} {width} {height}")
        if yolo_lines:
            label_file = img_file.replace(".png", ".txt")
            with open(os.path.join(label_dir, label_file), "w") as out:
                out.write("\n".join(yolo_lines))

# Auto-label train/val/test splits
auto_label_images(os.path.join(dst_images, "train"), os.path.join(dst_labels, "train"))
auto_label_images(os.path.join(dst_images, "val"), os.path.join(dst_labels, "val"))
auto_label_images(os.path.join(dst_images, "test"), os.path.join(dst_labels, "test"))

print("✅ Auto-labeling complete with MobileNetV3")

# -----------------------------
# Step 4: Create data.yaml
# -----------------------------
yaml_content = f"""
train: "D:/adas/images/train"
val: "D:/adas/images/val"
test: "D:/adas/images/test"

nc: {len(classes)}
names: {classes}
"""

with open(r"D:/adas/data.yaml", "w") as f:
    f.write(yaml_content)

print("✅ data.yaml created")
```

```python
transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225])
])

BATCH_SIZE = 32
LEARNING_RATE = 1e-3
EPOCHS = 50
IMG_SIZE = 640

# Optimizer setup
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.models as models

# Load MobileNetV3 classifier
classifier = models.mobilenet_v3_small(pretrained=True)
classifier.classifier[3] = nn.Linear(classifier.classifier[3].in_features, 12)  # 12 Indian road classes

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(classifier.parameters(), lr=LEARNING_RATE)
```

```python
from ultralytics import YOLO


# Train YOLOv8
model = YOLO("yolov8m.pt")   # medium model
model.train(data="D:\adas\data.yaml",
            epochs=EPOCHS,
            imgsz=IMG_SIZE,
            batch=BATCH_SIZE,
            lr0=LEARNING_RATE,
            optimizer="Adam")
```
